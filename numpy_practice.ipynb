```python
import numpy as np

# Creating arrays
arr1 = np.array([1, 2, 3, 4, 5])

print("Array:")
print(arr1)

# Mathematical operations
print("Multiply by 2:")
print(arr1 * 2)

print("Addition:")
print(arr1 + 10)

# Mean value
print("Mean:", np.mean(arr1))

# Maximum value
print("Max:", np.max(arr1))
```
